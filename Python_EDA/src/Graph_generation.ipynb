```python
# ----------------------------------------------------------
# EV POPULATION BY MODEL YEAR
#
# What it shows:
# The number of registered EVs belonging to each model year.
#
# Why it is important:
# Helps understand how the EV population is distributed across
# older and newer vehicle generations and gives an indication
# of how strongly newer EVs are represented in the market.
# ----------------------------------------------------------

year_counts = df["Model Year"].value_counts().sort_index()

plt.figure(figsize=(12, 6))

plt.plot(
    year_counts.index,
    year_counts.values,
    marker="o"
)

plt.title("EV Population by Model Year")
plt.xlabel("Model Year")
plt.ylabel("Number of Vehicles")
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "ev_population_by_model_year.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ----------------------------------------------------------
# TOP 10 EV MANUFACTURERS
#
# What it shows:
# The manufacturers with the largest number of registered EVs.
#
# Why it is important:
# Helps an EV company understand the competitive landscape,
# identify dominant manufacturers, and compare market presence.
# ----------------------------------------------------------

top_makes = df["Make"].value_counts().head(10).sort_values()

plt.figure(figsize=(10, 6))

plt.barh(
    top_makes.index,
    top_makes.values
)

plt.title("Top 10 EV Manufacturers")
plt.xlabel("Number of Vehicles")
plt.ylabel("Manufacturer")

plt.savefig(
    GRAPH_DIR / "top_10_manufacturers.png",
    dpi=300,
    bbox_inches="tight"
)

plt.tight_layout()
plt.show()
```

```python
# ----------------------------------------------------------
# TOP 10 EV MODELS
#
# What it shows:
# The individual vehicle models that are most common in
# the EV population.
#
# Why it is important:
# Manufacturer-level analysis tells us which companies are
# successful, while model-level analysis shows which actual
# products are most popular among customers.
# ----------------------------------------------------------

df["Vehicle"] = df["Make"].astype(str) + " " + df["Model"].astype(str)

top_models = (
    df["Vehicle"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_models.columns = ["Vehicle", "Count"]

plt.figure(figsize=(11, 6))

sns.barplot(
    data=top_models,
    x="Count",
    y="Vehicle"
)

plt.title("Top 10 EV Models")
plt.xlabel("Number of Vehicles")
plt.ylabel("Vehicle")

plt.tight_layout()

plt.savefig( 
            GRAPH_DIR / "top_10_models.png",
             dpi = 300,
             bbox_inches = "tight"
            )

plt.show()
```

```python
# ----------------------------------------------------------
# BEV VS PHEV DISTRIBUTION
#
# What it shows:
# The proportion of Battery Electric Vehicles (BEVs) compared
# with Plug-in Hybrid Electric Vehicles (PHEVs).
#
# Why it is important:
# Helps understand which EV technology currently has the
# stronger presence in the market.
#
# This can help manufacturers decide whether market preference
# is moving toward full electrification or hybrid solutions.
# ----------------------------------------------------------

ev_type_counts = df["Electric Vehicle Type"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    ev_type_counts.values,
    labels=ev_type_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.55}
)

plt.title("BEV vs PHEV Distribution")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "bev_vs_phev.png",
    dpi = 300,
    bbox_inches = "tight"
)

plt.show()
```

```python
# ----------------------------------------------------------
# BEV VS PHEV ACROSS MODEL YEARS
#
# What it shows:
# How the number of BEVs and PHEVs differs across model years.
#
# Why it is important:
# The overall BEV/PHEV chart only shows the current composition.
#
# This graph helps determine whether newer model years are
# increasingly dominated by one electric vehicle technology.
# ----------------------------------------------------------

ev_type_year = pd.crosstab(
    df["Model Year"],
    df["Electric Vehicle Type"]
)

ev_type_year = ev_type_year[ev_type_year.index >= 2015]

ev_type_year.plot(
    kind="bar",
    stacked=True,
    figsize=(13, 7)
)

plt.title("BEV vs PHEV Across Model Years")
plt.xlabel("Model Year")
plt.ylabel("Number of Vehicles")

plt.xticks(rotation=45)
plt.legend(title="Electric Vehicle Type")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "bev_phev_by_year.png",
    dpi = 300,
    bbox_inches = "tight"   
)

plt.show()
```

```python
# ----------------------------------------------------------
# TOP 10 COUNTIES BY EV POPULATION
#
# What it shows:
# The counties containing the largest number of registered EVs.
#
# Why it is important:
# Helps identify geographic areas with high EV concentration.
#
# An EV company could use this information when considering:
# - dealership locations
# - service centres
# - marketing campaigns
# - charging infrastructure
# ----------------------------------------------------------

top_counties = (
    df["County"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_counties.columns = ["County", "Count"]

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_counties,
    x="Count",
    y="County"
)

plt.title("Top 10 Counties by EV Population")
plt.xlabel("Number of Vehicles")
plt.ylabel("County")

plt.tight_layout()
plt.show()
```

```python
# ----------------------------------------------------------
# TOP 10 CITIES BY EV POPULATION
#
# What it shows:
# The cities with the highest number of registered EVs.
#
# Why it is important:
# Provides a more detailed view of geographic EV concentration
# than county-level analysis.
#
# This could help identify specific urban markets with
# particularly strong EV presence.
# ----------------------------------------------------------

top_cities = (
    df["City"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_cities.columns = ["City", "Count"]

plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_cities,
    x="Count",
    y="City"
)

plt.title("Top 10 Cities by EV Population")
plt.xlabel("Number of Vehicles")
plt.ylabel("City")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "top_10_cities.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ----------------------------------------------------------
# MANUFACTURER VS ELECTRIC VEHICLE TYPE
#
# What it shows:
# The number of BEVs and PHEVs produced by the most common
# manufacturers in the dataset.
#
# Why it is important:
# Helps understand competitor product strategies.
#
# For example, some manufacturers may primarily compete in
# full battery EVs while others may have a stronger presence
# in plug-in hybrids.
# ----------------------------------------------------------

top_10_makes = df["Make"].value_counts().head(10).index

make_type_data = df[df["Make"].isin(top_10_makes)]

make_type_table = pd.crosstab(
    make_type_data["Make"],
    make_type_data["Electric Vehicle Type"]
)

plt.figure(figsize=(10, 7))

sns.heatmap(
    make_type_table,
    annot=True,
    fmt="d",
    cmap="Reds"
)

plt.title("EV Type Distribution Among Top Manufacturers")
plt.xlabel("Electric Vehicle Type")
plt.ylabel("Manufacturer")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "manufacturer_ev_type_heatmap.png.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ----------------------------------------------------------
# MANUFACTURER PRESENCE ACROSS MODEL YEARS
#
# What it shows:
# How the number of vehicles from major manufacturers changes
# across model years.
#
# Why it is important:
# A simple manufacturer ranking shows who is largest overall.
#
# This graph gives additional competitive insight by showing
# which manufacturers have a stronger presence among newer
# vehicle generations.
# ----------------------------------------------------------

top_5_makes = df["Make"].value_counts().head(5).index

manufacturer_year = (
    df[df["Make"].isin(top_5_makes)]
    .groupby(["Model Year", "Make"])
    .size()
    .reset_index(name="Count")
)

manufacturer_year = manufacturer_year[
    manufacturer_year["Model Year"] >= 2015
]

plt.figure(figsize=(12, 7))

sns.lineplot(
    data=manufacturer_year,
    x="Model Year",
    y="Count",
    hue="Make",
    marker="o"
)

plt.title("Top Manufacturer Presence Across Model Years")
plt.xlabel("Model Year")
plt.ylabel("Number of Vehicles")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "manufacturer_trends.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
