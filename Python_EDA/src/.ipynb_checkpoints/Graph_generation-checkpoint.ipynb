### Below this - Data cleaning

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path
```

```python
GRAPH_DIR = Path(
    r"D:/Kanishka/@inventing/codinggg/Languages/Python/DVP/DVP Website+Presentation/EV-Population-Insights/website/graphs"
)

GRAPH_DIR.mkdir(parents=True, exist_ok=True)
```

```python
path = r"D:/Kanishka/@inventing/codinggg/Languages/Python/DVP/DVP Website+Presentation/EV-Population-Insights/data/Electric_Vehicle_Population_Data.csv"

df = pd.read_csv(path)

print ("Number of rows and columns")
print(df.shape)

print("FIrst 10 rows")
df.head(10)
```

```python
df.columns.tolist()

df.dtypes
```

```python
print("Misssing Values\n")
print(df.isnull().sum())

print("\nZeros\n")
zero_counts = (df == 0).sum()

print(zero_counts)
```

```python
df.describe()
```

```python
df.drop(
    columns=[
        "VIN (1-10)",
        "State",
        "DOL Vehicle ID",
        "Legislative District",
        "2020 Census Tract", 
        "Electric Range",
        "Base MSRP"        
    ],
    inplace=True
)

# vin, state, vehical id, census district and legislative arent useful to someone analysing the data for market 
# electric range and base msrp have too many zeros to be useful as columns
```

```python
df.columns.tolist()

df.dtypes
```

```python
df.isnull().sum()
```

```python
missing_location = df[
    df["County"].isna() |
    df["City"].isna() |
    df["Postal Code"].isna()
]

display(missing_location)
```

```python
df = df.dropna(subset=["County", "City", "Postal Code"]).copy()

# the ones with missing city and county cant ve derived for each others values so i dropped them
```

```python
df.isnull().sum()
```

```python
df = df.dropna(subset=["Vehicle Location"]).copy()
```

```python
df.isnull().sum()
```

### Below this - Graph generation

```python
# ----------------------------------------------------------
# EV POPULATION BY MODEL YEAR
#
# What it shows:
# The number of registered EVs belonging to each model year.
#
# Why it is important:
# Helps understand how the EV population is distributed across
# older and newer vehicle generations and gives an indication
# of how strongly newer EVs are represented in the market.
# ----------------------------------------------------------

year_counts = df["Model Year"].value_counts().sort_index()

plt.figure(figsize=(12, 6))

plt.plot(
    year_counts.index,
    year_counts.values,
    marker="o"
)

plt.title("EV Population by Model Year")
plt.xlabel("Model Year")
plt.ylabel("Number of Vehicles")
plt.grid(alpha=0.3)

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "ev_population_by_model_year.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# ----------------------------------------------------------
# TOP 10 EV MANUFACTURERS
#
# What it shows:
# The manufacturers with the largest number of registered EVs.
#
# Why it is important:
# Helps an EV company understand the competitive landscape,
# identify dominant manufacturers, and compare market presence.
# ----------------------------------------------------------

top_makes = df["Make"].value_counts().head(10).sort_values()

plt.figure(figsize=(10, 6))

plt.barh(
    top_makes.index,
    top_makes.values
)

plt.title("Top 10 EV Manufacturers")
plt.xlabel("Number of Vehicles")
plt.ylabel("Manufacturer")

plt.savefig(
    GRAPH_DIR / "top_10_manufacturers.png",
    dpi=300,
    bbox_inches="tight"
)

plt.tight_layout()
plt.show()
```

```python
# ----------------------------------------------------------
# TOP 10 EV MODELS
#
# What it shows:
# The individual vehicle models that are most common in
# the EV population.
#
# Why it is important:
# Manufacturer-level analysis tells us which companies are
# successful, while model-level analysis shows which actual
# products are most popular among customers.
# ----------------------------------------------------------

df["Vehicle"] = df["Make"].astype(str) + " " + df["Model"].astype(str)

top_models = (
    df["Vehicle"]
    .value_counts()
    .head(10)
    .reset_index()
)

top_models.columns = ["Vehicle", "Count"]

plt.figure(figsize=(11, 6))

sns.barplot(
    data=top_models,
    x="Count",
    y="Vehicle"
)

plt.title("Top 10 EV Models")
plt.xlabel("Number of Vehicles")
plt.ylabel("Vehicle")

plt.tight_layout()

plt.savefig( 
            GRAPH_DIR / "top_10_models.png",
             dpi = 300,
             bbox_inches = "tight"
            )

plt.show()
```

```python
# ----------------------------------------------------------
# BEV VS PHEV DISTRIBUTION
#
# What it shows:
# The proportion of Battery Electric Vehicles (BEVs) compared
# with Plug-in Hybrid Electric Vehicles (PHEVs).
#
# Why it is important:
# Helps understand which EV technology currently has the
# stronger presence in the market.
#
# This can help manufacturers decide whether market preference
# is moving toward full electrification or hybrid solutions.
# ----------------------------------------------------------

ev_type_counts = df["Electric Vehicle Type"].value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    ev_type_counts.values,
    labels=ev_type_counts.index,
    autopct="%1.1f%%",
    startangle=90,
    wedgeprops={"width": 0.55}
)

plt.title("BEV vs PHEV Distribution")

plt.tight_layout()

plt.savefig(
    GRAPH_DIR / "bev_vs_phev.png",
    dpi = 300,
    bbox_inches = "tight"
)

plt.show()
```
