```python
missing_location = df[
    df["County"].isna() |
    df["City"].isna() |
    df["Postal Code"].isna()
]

display(missing_location)
```

```python
df = df.dropna(subset=["County", "City", "Postal Code"]).copy()
```

```python
df = df.dropna(subset=["Vehicle Location"]).copy()
```

```python
df.loc[
    df["Electric Range"].isna() | df["Base MSRP"].isna(),
    ["Model Year", "Make", "Model", "Electric Vehicle Type",
     "Electric Range", "Base MSRP"]
]
```

```python
missing_rows = df[
    df["Electric Range"].isna() |
    df["Base MSRP"].isna()
]

for _, row in missing_rows.iterrows():

    print("\nMissing vehicle:")
    display(row[
        ["Model Year", "Make", "Model",
         "Electric Range", "Base MSRP"]
    ].to_frame().T)

    print("Matching vehicles:")

    matches = df[
        (df["Model Year"] == row["Model Year"]) &
        (df["Make"] == row["Make"]) &
        (df["Model"] == row["Model"])
    ]

    display(
        matches[
            ["Model Year", "Make", "Model",
             "Electric Range", "Base MSRP"]
        ]
        .drop_duplicates()
    )
```

```python
mask = (
    (df["Model Year"] == 2024) &
    (df["Make"] == "MERCEDES-BENZ") &
    (df["Model"] == "S-CLASS") &
    (df["Electric Range"].isna())
)

df.loc[mask, "Electric Range"] = 46
```
